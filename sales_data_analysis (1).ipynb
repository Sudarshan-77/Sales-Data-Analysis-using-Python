**pandas Q13** `Intermediate`  Create a column `OrderSize`: 'Small' (revenue below 5,000), 'Medium' (5,000 to 30,000) and 'Large' (above 30,000). Count the orders in each size.

```python


conditions = [
    df['Revenue'] < 5000,
    df['Revenue'].between(5000, 30000),
    df['Revenue'] > 30000
]

choice = ['Small', 'Medium', 'Large']

df["OrderSize"] = np.select(
    conditions,
    choice,
    default='Unknown'
)

print(df["OrderSize"].value_counts())
```

**Matplotlib Q9** `Intermediate`  Draw monthly revenue for Electronics and for all other categories as two lines. Add a legend and a grid, and mark the peak month with `plt.annotate`.

```python
import matplotlib.pyplot as plt

df['realdate'] = pd.to_datetime(df['realdate'])
df['Month'] = df['realdate'].dt.month

electronics = df[df['Category'] == 'Electronics'].groupby('Month')['Revenue'].sum()
others = df[df['Category'] != 'Electronics'].groupby('Month')['Revenue'].sum()

plt.figure(figsize=(10, 5))

plt.plot(electronics.index, electronics.values, marker='o', label='Electronics')
plt.plot(others.index, others.values, marker='o', label='Others')

peak = electronics.idxmax()

plt.annotate('Peak', (peak, electronics.max()))
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.legend()
plt.grid(True)
plt.show()
```
